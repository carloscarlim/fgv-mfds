## 2) Variáveis e tipos (o mínimo essencial)

Em Python, você cria uma variável com `nome = valor`.

Principais tipos que você vai ver o tempo todo:
- `int`: inteiro (ex.: 10)
- `float`: decimal (ex.: 10.5)
- `str`: texto (ex.: "Ana")
- `bool`: verdadeiro/falso (`True`/`False`)

Use `type()` para descobrir o tipo.

````python
idade = 35
renda = 8500.50
nome = "Ana"
ativo = True

type(idade), type(renda), type(nome), type(ativo)
````

## 3) Operações e textos (strings)

Operações comuns:
- `+` soma
- `-` subtração
- `*` multiplicação
- `/` divisão

Para criar textos com valores, usamos **f-strings** (`f"...{variavel}..."`).

````python
renda_anual = renda * 12
mensagem = f"{nome} tem renda anual de R$ {renda_anual:,.2f}"
mensagem
````

## 4) Listas (listas de valores)

Uma **lista** guarda vários itens em uma sequência.

Exemplos em dados:
- uma lista de notas
- uma lista de idades
- uma lista de valores de gasto

Funções úteis:
- `len(lista)`: tamanho
- `sum(lista)`: soma

````python
notas = [7.5, 8.0, 6.5, 9.0]
media = sum(notas) / len(notas)
media
````

## 5) Dicionários (chave → valor)

Um **dicionário** representa pares `chave: valor`.

Acesso: `dicionario["chave"]`

````python
cliente = {"nome": "Ana", "cidade": "SP", "plano": "Ouro"}
cliente["plano"]
````

## 6) Controle de fluxo (if/for) — versão mínima

Você vai usar `if` para **tomar decisões** e `for` para **repetir** algo.

> Observação: em análise de dados, muitas vezes usamos `pandas` para evitar loops. Aqui é só para você entender a lógica.

````python
gastos = [120, 80, 450, 60, 999]
altos = []

for g in gastos:
    if g > 300:
        altos.append(g)

altos
````

## 7) Funções (para reutilizar lógica)

Uma função evita repetir código.

Formato:
```python
def nome(parametro):
    ...
    return resultado
```

Vamos criar uma função que classifica gasto em **baixo/médio/alto**.

````python
def classificar_gasto(valor):
    if valor >= 300:
        return "alto"
    elif valor >= 100:
        return "médio"
    else:
        return "baixo"

classificar_gasto(450), classificar_gasto(90), classificar_gasto(150)
````

## 9) Criando um dataset didático (sem baixar arquivo)

Para a aula ser **autocontida**, vamos criar um conjunto de dados pequeno.

Ele simula um cadastro de clientes com:
- colunas numéricas (idade, renda, gasto)
- colunas categóricas (cidade, plano, canal)
- **faltantes** (`NaN`) e **outliers** (valores muito altos)

Isso é ótimo porque nas próximas aulas vamos tratar esses problemas.

````python
np.random.seed(42)

df = pd.DataFrame({
    "idade": np.random.randint(18, 70, 20),
    "renda_mensal": np.random.normal(6000, 2000, 20).round(0),
    "tempo_cliente_meses": np.random.randint(1, 120, 20),
    "gasto_mensal": np.random.normal(350, 120, 20).round(0),
    "cidade": np.random.choice(["SP", "RJ", "BH"], 20),
    "plano": np.random.choice(["Bronze", "Prata", "Ouro"], 20),
    "canal": np.random.choice(["app", "loja", "site"], 20)
})

# Inserindo faltantes
df.loc[[3, 7], "renda_mensal"] = np.nan
df.loc[[5], "gasto_mensal"] = np.nan
df.loc[[12], "cidade"] = np.nan

# Inserindo outliers
df.loc[1, "gasto_mensal"] = 2000
df.loc[9, "renda_mensal"] = 25000

df.head()
````

## 12) Frequência de categorias (value_counts)

## 14) Filtrar linhas (condições)

Filtros comuns:
- `df[df["col"] > valor]`
- combinações com `&` (E) e `|` (OU)

> Atenção: use parênteses em cada condição.

````python
df[df["plano"] == "Ouro"].head()
````

````python
df[(df["cidade"] == "SP") & (df["gasto_mensal"] > 400)]
````

## 15) GroupBy: agregações por grupo (superpoder do pandas)

`groupby` permite responder perguntas do tipo:
- gasto médio por plano
- renda média por cidade

Exemplo:
```python
df.groupby("plano")["gasto_mensal"].mean()
```

````python
df.groupby("plano")["gasto_mensal"].mean()
````

````python
df.groupby(["cidade", "plano"])["gasto_mensal"].mean()
````

## 16) Criando uma coluna nova (feature engineering simples)

Vamos usar a função `classificar_gasto()` que criamos antes.

`apply()` aplica a função linha a linha na coluna.

````python
df["gasto_categoria"] = df["gasto_mensal"].apply(lambda x: classificar_gasto(x) if pd.notna(x) else "desconhecido")
df[["gasto_mensal", "gasto_categoria"]].head(10)
````

## 17) Gráfico rápido: histograma

O histograma ajuda a ver:
- a distribuição dos dados
- valores muito extremos (possíveis outliers)

Vamos visualizar `gasto_mensal`.

````python
import matplotlib.pyplot as plt

df["gasto_mensal"].hist(bins=10)
plt.title("Distribuição do gasto mensal")
plt.xlabel("gasto_mensal")
plt.ylabel("frequência")
plt.show()
````
