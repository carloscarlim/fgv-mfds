## 5. Repetições com `for`

## 6. Funções: uma regra de negócio reutilizável
Uma **função** empacota um cálculo com um nome. Você define uma vez (`def`) e usa quantas vezes quiser.
Esta é a fórmula da Calculadora de Valor e Risco da Aula 1:

```python
def valor_liquido(volume, acerto, ganho, perda, custo_decisao, custo_fixo, risco):
    """Valor líquido anual de uma aplicação de CD (modelo da Aula 1). acerto entre 0 e 1."""
    por_decisao = acerto * ganho - (1 - acerto) * perda - custo_decisao
    return volume * por_decisao - custo_fixo - risco

v = valor_liquido(volume=20000, acerto=0.70, ganho=900, perda=250, custo_decisao=1, custo_fixo=2_500_000, risco=300_000)
print(f"Valor líquido: R$ {v:,.0f}".replace(",", "."))
```

## 8. Demonstração do professor: aonde isso leva
> **Só assista.** Você não precisa entender cada linha agora. As técnicas aparecem com calma nas próximas aulas e nas disciplinas seguintes do MBA.

Base: **Olist**, e-commerce brasileiro com cerca de 100 mil pedidos (2016–2018), disponibilizada publicamente pela empresa no Kaggle.
Preparamos um resumo por cliente com: número de pedidos, valor gasto, dias desde a última compra, atraso na entrega e nota da avaliação.

```python
clientes = pd.read_csv("https://raw.githubusercontent.com/carloscarlim/fgv-mfds/main/dados/olist_demo_clientes.csv.gz")
print(clientes.shape)
clientes.head()
```

### 8.1 Agrupamento (não supervisionado): que tipos de clientes temos?
O algoritmo **k-means** forma grupos de clientes parecidos sem saber nada de antemão sobre eles.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

X = clientes[["recencia_dias", "valor_total", "atraso_dias"]]
X_padronizado = StandardScaler().fit_transform(X)        # coloca as variáveis na mesma escala
clientes["grupo"] = KMeans(n_clusters=4, n_init=10, random_state=42).fit_predict(X_padronizado)

clientes.groupby("grupo")[["recencia_dias", "valor_total", "atraso_dias", "nota"]].mean().round(1).assign(clientes=clientes["grupo"].value_counts())
```

**Para discutir:** que nome de negócio você daria a cada grupo? Que ação faria com cada um?

### 8.2 Classificação (supervisionada): o atraso prevê uma avaliação ruim?
Agora temos uma **resposta conhecida**: a nota que cada cliente deu. Vamos treinar um modelo para prever se a nota será ruim (1 ou 2) e testá-lo em clientes que ele nunca viu.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, recall_score, precision_score

clientes["nota_ruim"] = clientes["nota"] <= 2
X = clientes[["atraso_dias", "valor_total", "frete_medio"]]
y = clientes["nota_ruim"]
X_treino, X_teste, y_treino, y_teste = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

modelo = LogisticRegression(class_weight="balanced", max_iter=1000).fit(X_treino, y_treino)
previsto = modelo.predict(X_teste)

print(f"Clientes com nota ruim na base: {y.mean():.1%}")
print(f"Acurácia de quem sempre diz 'nota boa': {1 - y_teste.mean():.1%}")
print(f"Acurácia do modelo: {accuracy_score(y_teste, previsto):.1%}")
print(f"Recall (notas ruins que o modelo pegou): {recall_score(y_teste, previsto):.1%}")
print(f"Precisão (alertas que eram mesmo nota ruim): {precision_score(y_teste, previsto):.1%}")
```

## Erros comuns e como resolver

| Mensagem | O que aconteceu | O que fazer |
|---|---|---|
| `SyntaxError: expected ':'` | Faltaram os dois-pontos no `if`, `for` ou `def` | Coloque `:` no fim da linha |
| `IndentationError` | O recuo está errado | Use 4 espaços dentro de `if`, `for` e `def` |
| `TypeError: can only concatenate str (not "float") to str` | Tentou somar texto com número | Use `f"texto {numero}"` ou `str(numero)` |
| `NameError: name 'x' is not defined` | Variável não criada ou célula não executada | Execute as células anteriores |
| `KeyError: 'valor'` | Chave inexistente no dicionário | Confira o nome exato da chave (maiúsculas e acentos contam) |
| `IndexError: list index out of range` | Posição maior que o tamanho da lista | Lembre que a contagem começa em 0 |
