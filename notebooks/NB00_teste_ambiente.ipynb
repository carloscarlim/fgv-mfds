## 2. Carregar dados pela internet
O notebook tenta baixar um arquivo do repositório da disciplina. Se a sua rede bloquear o GitHub, ele usa uma cópia local e avisa.

```python
# Endereço do arquivo no repositório da disciplina
URL = 'https://raw.githubusercontent.com/carloscarlim/fgv-mfds/main/dados/teste_ambiente.csv'

try:
    df = pd.read_csv(URL)
    print('Download OK: sua rede acessa o repositório da disciplina.')
except Exception as erro:
    print('⚠️ Não consegui baixar o arquivo. Provável bloqueio de rede (comum em notebook corporativo).')
    print('Usando dados locais para continuar o teste. Avise no fórum "Setup" do ECLASS.')
    rng = np.random.default_rng(2026)
    df = pd.DataFrame({
        'id': range(1, 201),
        'regiao': rng.choice(['Norte', 'Nordeste', 'Centro-Oeste', 'Sudeste', 'Sul'], 200),
        'valor_pedido': rng.lognormal(4.5, 0.5, 200).round(2),
        'dias_entrega': np.clip(rng.normal(9, 4, 200).astype(int), 1, None),
    })

df.head()
```

## 3. Um resumo e um gráfico

```python
# Média de dias de entrega por região
print(df.groupby('regiao')['dias_entrega'].mean().round(1))

# Histograma do valor dos pedidos
sns.histplot(data=df, x='valor_pedido', bins=20)
plt.title('Distribuição do valor dos pedidos (dados fictícios)')
plt.xlabel('Valor do pedido (R$)')
plt.ylabel('Quantidade de pedidos')
plt.show()
```
