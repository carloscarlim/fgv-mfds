# Parte 1 · Rede Lar & Cia: mais vendedores, mais vendas?
O diretor comercial mostrou um gráfico na reunião: **lojas com mais vendedores vendem muito mais**. Ele quer contratar **2 vendedores a mais em cada uma das 200 lojas**: 400 contratações, cerca de **R\$ 24 milhões por ano** (custo hipotético de R\$ 5 mil por vendedor por mês). A pergunta: **contratar vai aumentar as vendas?**

| Coluna | O que é |
|---|---|
| `loja` | código da loja |
| `porte` | pequena, média ou grande |
| `vendedores` | vendedores na loja |
| `vendas_mil` | vendas mensais, em R\$ mil |

**O que acontece na célula abaixo**

- Lê a base e mostra as 5 primeiras linhas.

```python
lojas = pd.read_csv(RAIZ + "aula05_c1_lojas_vendedores.csv")
lojas.head()
```

**O que a diretoria viu**

**O que acontece na célula abaixo**

- `.corr()` calcula a correlação de Pearson entre vendedores e vendas, com **todas** as lojas.
- `plt.scatter(...)` desenha um ponto por loja.

```python
print("Correlação vendedores × vendas, todas as lojas:", round(lojas["vendedores"].corr(lojas["vendas_mil"]), 2))

plt.scatter(lojas["vendedores"], lojas["vendas_mil"], alpha=0.6)
plt.xlabel("vendedores na loja"); plt.ylabel("vendas mensais (R$ mil)")
plt.show()
```

**O que olhar na saída**

A correlação é alta? O que o gráfico parece dizer? Agora pense: o que faz uma loja ter, **ao mesmo tempo**, mais vendedores e mais vendas?

**O teste: comparar lojas do mesmo porte**

**O que acontece na célula abaixo**

- `groupby("porte")` separa as lojas por porte; para cada porte, `.mean()` mostra a média de vendedores e de vendas.
- O `for` calcula a correlação **dentro** de cada porte.
- No gráfico, cada porte ganha uma cor.

```python
print(lojas.groupby("porte")[["vendedores", "vendas_mil"]].mean())
print()
for porte, grupo in lojas.groupby("porte"):
    print(f"Correlação dentro das lojas de porte {porte}: {grupo['vendedores'].corr(grupo['vendas_mil']):.2f}")

for porte, grupo in lojas.groupby("porte"):
    plt.scatter(grupo["vendedores"], grupo["vendas_mil"], alpha=0.6, label=porte)
plt.xlabel("vendedores na loja"); plt.ylabel("vendas mensais (R$ mil)"); plt.legend()
plt.show()
```

**O que olhar na saída**

Compare a correlação de todas as lojas com a correlação **dentro** de cada porte. O que sobra da relação quando as lojas são parecidas? No gráfico, as três nuvens sobem ou ficam "deitadas"?

**O que acontece na célula abaixo**

- Este é o gráfico para o slide, já no formato do guia: **um gráfico, uma mensagem**; cinza para o contexto e **uma cor** para o que importa; o número escrito no lugar onde é lido; sem grade e sem bordas.
- **Troque o texto de `titulo`** pela sua Big Idea e execute de novo. Para levar ao slide: clique com o botão direito na imagem e copie (ou salve).

```python
titulo = "Escreva aqui a sua Big Idea"

r_todas = lojas["vendedores"].corr(lojas["vendas_mil"])
r_porte = [grupo["vendedores"].corr(grupo["vendas_mil"]) for porte, grupo in lojas.groupby("porte")]
nomes = ["todas as lojas"] + ["só " + p + "s" for p, g in lojas.groupby("porte")]
valores = [r_todas] + r_porte

fig, ax = plt.subplots(figsize=(8, 4))
barras = ax.bar(nomes, valores, color=["#A9B5C7"] + ["#0E6FA6"] * 3)
for b, v in zip(barras, valores):
    ax.text(b.get_x() + b.get_width() / 2, max(v, 0) + 0.03, f"{v:.2f}".replace(".", ","), ha="center", fontsize=13, fontweight="bold")
ax.set_title(titulo, loc="left", fontsize=14, fontweight="bold")
ax.set_ylabel("correlação vendedores × vendas"); ax.set_ylim(-0.3, 1.05)
ax.axhline(0, color="#4A5568", linewidth=0.8)
for lado in ["top", "right"]:
    ax.spines[lado].set_visible(False)
plt.show()
```

# Parte 2 · Farma Mais: mais segurança, mais furtos?
A auditoria cruzou o **investimento em segurança** de cada uma das 150 farmácias com os **furtos do ano**: as lojas que mais investem são as que mais sofrem furtos. O financeiro propõe **cortar o investimento** (cerca de **R\$ 11,6 milhões por ano**), "porque não funciona". A pergunta: **segurança não reduz furtos?**

| Coluna | O que é |
|---|---|
| `loja` | código da farmácia |
| `furtos_ano1` | furtos no ano anterior ao investimento |
| `investimento_ano2_mil` | investimento em segurança no ano 2, em R\$ mil |
| `furtos_ano2` | furtos no ano 2 |

**O que acontece na célula abaixo**

- Lê a base e mostra as 5 primeiras linhas.

```python
farmacias = pd.read_csv(RAIZ + "aula05_c2_lojas_seguranca.csv")
farmacias.head()
```

**O que a diretoria viu: a foto do ano 2**

**O que acontece na célula abaixo**

- Correlação entre o investimento do ano 2 e os furtos do ano 2, e o gráfico de dispersão.

```python
print("Correlação investimento × furtos do ano 2:", round(farmacias["investimento_ano2_mil"].corr(farmacias["furtos_ano2"]), 2))

plt.scatter(farmacias["investimento_ano2_mil"], farmacias["furtos_ano2"], alpha=0.6)
plt.xlabel("investimento em segurança no ano 2 (R$ mil)"); plt.ylabel("furtos no ano 2")
plt.show()
```

**O que olhar na saída**

A correlação é positiva? Antes de concluir, pergunte: **o que veio primeiro**, o furto ou o investimento? Quem decidiu quanto cada loja ia investir, e com base em quê?

**O teste: cada loja antes e depois do investimento**

**O que acontece na célula abaixo**

- `variacao` é a mudança dos furtos de cada loja, do ano 1 para o ano 2, em %.
- `pd.qcut(..., 3, ...)` divide as lojas em três grupos do mesmo tamanho pelo investimento: o terço que menos investiu, o do meio e o que mais investiu.
- A tabela mostra, para cada terço, os furtos médios nos dois anos e a variação média.

```python
farmacias["variacao"] = (farmacias["furtos_ano2"] - farmacias["furtos_ano1"]) / farmacias["furtos_ano1"] * 100
farmacias["grupo"] = pd.qcut(farmacias["investimento_ano2_mil"], 3, labels=["investiu menos", "meio", "investiu mais"])

print("Correlação investimento × variação dos furtos:", round(farmacias["investimento_ano2_mil"].corr(farmacias["variacao"]), 2))
farmacias.groupby("grupo", observed=True)[["furtos_ano1", "furtos_ano2", "variacao"]].mean()
```

**O que olhar na saída**

Quem investiu mais tinha mais ou menos furtos **antes** (ano 1)? E quanto os furtos **caíram** em cada terço? A correlação com a variação tem o mesmo sinal da foto do ano?

**O que acontece na célula abaixo**

- Este é o gráfico para o slide, já no formato do guia: **um gráfico, uma mensagem**; cinza para o contexto e **uma cor** para o que importa; o número escrito no lugar onde é lido; sem grade e sem bordas.
- **Troque o texto de `titulo`** pela sua Big Idea e execute de novo. Para levar ao slide: clique com o botão direito na imagem e copie (ou salve).

```python
titulo = "Escreva aqui a sua Big Idea"

tabela = farmacias.groupby("grupo", observed=True)["variacao"].mean()

fig, ax = plt.subplots(figsize=(8, 4))
barras = ax.bar(tabela.index, tabela.values, color=["#A9B5C7", "#A9B5C7", "#0E6FA6"])
for b, v in zip(barras, tabela.values):
    ax.text(b.get_x() + b.get_width() / 2, v / 2, f"{v:.0f}%", ha="center", color="white", fontsize=14, fontweight="bold")
ax.set_title(titulo, loc="left", fontsize=14, fontweight="bold")
ax.set_ylabel("variação dos furtos, ano 1 → ano 2 (%)")
ax.axhline(0, color="#4A5568", linewidth=0.8)
for lado in ["top", "right", "bottom"]:
    ax.spines[lado].set_visible(False)
plt.show()
```

# Parte 3 · Café Bom Clube: cliente antigo gasta mais?
Na base de **assinantes ativos**, quem é cliente há mais tempo gasta mais por mês. O marketing quer lançar um **programa de fidelidade** caro (cerca de **R\$ 2 milhões**, custo hipotético), porque "o tempo de casa faz o cliente gastar mais". A pergunta: **o tempo de casa aumenta o gasto?**

| Coluna | O que é |
|---|---|
| `assinante` | código do assinante |
| `mes_entrada` | mês em que assinou (1 a 36; hoje é o mês 36) |
| `mes_cancelamento` | mês em que cancelou (vazio se continua ativo) |
| `gasto_mensal` | gasto por mês, em R\$, no plano que o assinante escolheu (não muda com o tempo) |

**O que acontece na célula abaixo**

- Lê a base e mostra as 5 primeiras linhas.

```python
assinantes = pd.read_csv(RAIZ + "aula05_c3_assinantes.csv")
assinantes.head()
```

**O que a diretoria viu: os ativos de hoje**

**O que acontece na célula abaixo**

- `ativos` fica só com quem **não** cancelou (`mes_cancelamento` vazio: `.isna()`).
- `tempo_casa` = meses desde a entrada; `pd.cut` cria três faixas de tempo de casa.
- A tabela mostra o gasto médio e o número de assinantes em cada faixa.

```python
ativos = assinantes[assinantes["mes_cancelamento"].isna()].copy()
ativos["tempo_casa"] = 37 - ativos["mes_entrada"]
ativos["faixa"] = pd.cut(ativos["tempo_casa"], bins=[0, 12, 24, 36], labels=["até 1 ano", "1 a 2 anos", "mais de 2 anos"])

ativos.groupby("faixa", observed=True)["gasto_mensal"].agg(["mean", "count"])
```

**O que olhar na saída**

O gasto sobe com o tempo de casa? Agora repare na contagem: por que há **tão poucos** ativos com mais de 2 anos? Para onde foram os outros clientes que entraram há tanto tempo?

**O teste: acompanhar quem entrou no primeiro ano**

**O que acontece na célula abaixo**

- `safra` = todos os que entraram nos 12 primeiros meses, **incluindo quem cancelou**.
- `ficou` marca com `True` quem continua ativo; a tabela compara o gasto de quem ficou e de quem cancelou.

```python
safra = assinantes[assinantes["mes_entrada"] <= 12].copy()
safra["ficou"] = safra["mes_cancelamento"].isna()

print("Assinantes que entraram no 1º ano:", len(safra))
print(f"Ainda ativos hoje: {safra['ficou'].mean():.0%}")
safra.groupby("ficou")["gasto_mensal"].agg(["mean", "count"])
```

**O que olhar na saída**

Quanto gastava, **desde a entrada**, quem cancelou? E quem ficou? Lembre: o gasto de cada assinante **não muda** com o tempo. Então, o que faz a média dos antigos ser maior?

**O que acontece na célula abaixo**

- Este é o gráfico para o slide, já no formato do guia: **um gráfico, uma mensagem**; cinza para o contexto e **uma cor** para o que importa; o número escrito no lugar onde é lido; sem grade e sem bordas.
- **Troque o texto de `titulo`** pela sua Big Idea e execute de novo. Para levar ao slide: clique com o botão direito na imagem e copie (ou salve).

```python
titulo = "Escreva aqui a sua Big Idea"

nomes = ["todos que entraram\nno 1º ano", "cancelaram", "continuam ativos"]
valores = [safra["gasto_mensal"].mean(),
           safra.loc[~safra["ficou"], "gasto_mensal"].mean(),
           safra.loc[safra["ficou"], "gasto_mensal"].mean()]

fig, ax = plt.subplots(figsize=(8, 4))
barras = ax.bar(nomes, valores, color=["#A9B5C7", "#0E6FA6", "#A9B5C7"])
for b, v in zip(barras, valores):
    ax.text(b.get_x() + b.get_width() / 2, v + 2, f"R$ {v:.0f}", ha="center", fontsize=13, fontweight="bold")
ax.set_title(titulo, loc="left", fontsize=14, fontweight="bold")
ax.set_ylabel("gasto mensal médio (R$)")
for lado in ["top", "right", "left"]:
    ax.spines[lado].set_visible(False)
ax.set_yticks([])
plt.show()
```

# Parte 4 · Rede Ômega: o piloto nas piores lojas
A rede escolheu as **300 lojas com as piores vendas** do trimestre para um treinamento. Por falta de vagas, **sorteou 150** delas para fazer o treinamento agora; as outras 150 ficaram para o semestre seguinte. No trimestre seguinte, as lojas treinadas **melhoraram muito**, e o RH quer expandir o treinamento para as **2.000 lojas** (cerca de **R\$ 20 milhões**, custo hipotético). A pergunta: **o treinamento funcionou?**

| Coluna | O que é |
|---|---|
| `loja` | código da loja |
| `vendas_t1`, `vendas_t2` | índice de vendas no trimestre 1 e no trimestre 2 (100 = média da rede) |
| `escolhida` | `True` para as 300 piores do trimestre 1 |
| `treinada` | `True` para as 150 sorteadas entre as escolhidas |

**O que acontece na célula abaixo**

- Lê a base e cria a coluna `melhora` (trimestre 2 − trimestre 1).

```python
omega = pd.read_csv(RAIZ + "aula05_c4_lojas_piloto.csv")
omega["melhora"] = omega["vendas_t2"] - omega["vendas_t1"]
omega.head()
```

**O que a diretoria viu: as lojas treinadas**

**O que acontece na célula abaixo**

- `omega[omega["treinada"]]` fica só com as lojas treinadas; `.mean()` dá a melhora média.

```python
print("Melhora média das lojas treinadas:", round(omega[omega["treinada"]]["melhora"].mean(), 1), "pontos")
```

**O que olhar na saída**

As treinadas melhoraram? Antes de dar o crédito ao treinamento, pense: essas lojas foram escolhidas **por estar no pior trimestre**. Um trimestre muito ruim tem um pouco de azar. O azar se repete?

**O teste: o grupo que o sorteio deixou de fora**

**O que acontece na célula abaixo**

- `grupo` dá um nome a cada loja: treinada, escolhida mas **não** treinada (o controle sorteado) e as demais.
- A tabela mostra a melhora média e o número de lojas em cada grupo.

```python
omega["grupo"] = "demais lojas"
omega.loc[omega["escolhida"], "grupo"] = "piores, sem treinamento"
omega.loc[omega["treinada"], "grupo"] = "piores, com treinamento"

omega.groupby("grupo")["melhora"].agg(["mean", "count"])
```

**O que olhar na saída**

Compare as piores **com** e **sem** treinamento: as duas foram escolhidas do mesmo jeito, e só o sorteio as separou. Qual é a diferença entre elas? E as demais lojas, o que fizeram?

**O que acontece na célula abaixo**

- Este é o gráfico para o slide, já no formato do guia: **um gráfico, uma mensagem**; cinza para o contexto e **uma cor** para o que importa; o número escrito no lugar onde é lido; sem grade e sem bordas.
- **Troque o texto de `titulo`** pela sua Big Idea e execute de novo. Para levar ao slide: clique com o botão direito na imagem e copie (ou salve).

```python
titulo = "Escreva aqui a sua Big Idea"

tabela = omega.groupby("grupo")["melhora"].mean()
ordem = ["piores, com treinamento", "piores, sem treinamento", "demais lojas"]
valores = [tabela[g] for g in ordem]

fig, ax = plt.subplots(figsize=(8, 4))
barras = ax.bar(ordem, valores, color=["#0E6FA6", "#0E6FA6", "#A9B5C7"])
for b, v in zip(barras, valores):
    ax.text(b.get_x() + b.get_width() / 2, v + 0.6 if v > 0 else v - 1.8, f"{v:+.1f}".replace(".", ","), ha="center", fontsize=13, fontweight="bold")
ax.set_title(titulo, loc="left", fontsize=14, fontweight="bold")
ax.set_ylabel("variação do índice de vendas, T1 → T2"); ax.set_ylim(-5, 17)
ax.axhline(0, color="#4A5568", linewidth=0.8)
for lado in ["top", "right", "bottom"]:
    ax.spines[lado].set_visible(False)
plt.show()
```

# Parte 5 · Lojas Delta: o indicador que prevê as vendas
Um analista cruzou **24 meses de vendas** com **100 indicadores de mercado** e encontrou vários que "andam junto" com as vendas. A diretoria quer usar o **melhor deles** para planejar as compras do próximo ano. A pergunta: **o indicador com a maior correlação serve para prever as vendas?**

| Coluna | O que é |
|---|---|
| `mes` | mês, de 1 a 24 |
| `vendas` | índice de vendas do mês |
| `ind_001` a `ind_100` | os 100 indicadores de mercado |

**O que acontece na célula abaixo**

- Lê a base e mostra as 5 primeiras linhas (são 102 colunas: o pandas mostra só algumas).

```python
delta = pd.read_csv(RAIZ + "aula05_c5_indicadores.csv")
delta.head()
```

**O que a diretoria viu: os indicadores que andam junto**

**O que acontece na célula abaixo**

- `delta.drop(columns="mes").corr()["vendas"]` calcula a correlação de cada coluna com as vendas; `.drop("vendas")` tira a correlação das vendas com elas mesmas.
- `(r.abs() > 0.5).sum()` conta quantos indicadores têm correlação forte (acima de 0,5 ou abaixo de −0,5).
- `r.abs().idxmax()` acha o nome do indicador com a maior correlação.

```python
r = delta.drop(columns="mes").corr()["vendas"].drop("vendas")
melhor = r.abs().idxmax()
print("Indicadores com correlação forte (|r| > 0,5):", (r.abs() > 0.5).sum(), "de 100")
print("O melhor:", melhor, "com r =", round(r[melhor], 2))

fig, ax1 = plt.subplots(figsize=(8, 4))
ax1.plot(delta["mes"], delta["vendas"], color="#213B75", label="vendas")
ax2 = ax1.twinx()
ax2.plot(delta["mes"], delta[melhor], color="#0E6FA6", linestyle="--", label=melhor)
ax1.set_xlabel("mês"); ax1.set_ylabel("vendas"); ax2.set_ylabel(melhor)
plt.show()
```

**O que olhar na saída**

Quantos indicadores "andam junto" com as vendas? O melhor parece mesmo acompanhar a linha das vendas. Agora pense: as vendas **crescem com o tempo**. O que acontece com qualquer série que também sobe (ou desce) com o tempo?

**O teste: comparar as variações de um mês para o outro**

**O que acontece na célula abaixo**

- `.diff()` troca cada valor pela **variação** em relação ao mês anterior: tira a tendência e pergunta "quando as vendas sobem mais que o normal, o indicador também sobe?".
- As mesmas contas, agora com as variações.

```python
rv = delta.drop(columns="mes").diff().corr()["vendas"].drop("vendas")
print("Com as variações mensais:")
print("  indicadores com correlação forte:", (rv.abs() > 0.5).sum(), "de 100")
print("  o 'melhor' de antes,", melhor, ", agora tem r =", round(rv[melhor], 2))
```

**O que olhar na saída**

Quantos indicadores continuam fortes quando a tendência sai? E o "melhor" de antes? O que isso diz sobre usá-lo para planejar as compras?

**O que acontece na célula abaixo**

- Este é o gráfico para o slide, já no formato do guia: **um gráfico, uma mensagem**; cinza para o contexto e **uma cor** para o que importa; o número escrito no lugar onde é lido; sem grade e sem bordas.
- **Troque o texto de `titulo`** pela sua Big Idea e execute de novo. Para levar ao slide: clique com o botão direito na imagem e copie (ou salve). O gráfico mostra a correlação de cada indicador com as vendas, nos níveis (à esquerda) e nas variações (à direita).

```python
titulo = "Escreva aqui a sua Big Idea"

fig, ax = plt.subplots(1, 2, figsize=(11, 4), sharey=True)
ax[0].hist(r, bins=20, range=(-1, 1), color="#0E6FA6")
ax[0].set_title(f"Níveis: {(r.abs() > 0.5).sum()} de 100 'fortes'", loc="left")
ax[1].hist(rv, bins=20, range=(-1, 1), color="#A9B5C7")
ax[1].set_title(f"Variações mensais: {(rv.abs() > 0.5).sum()} de 100", loc="left")
for a in ax:
    a.axvline(-0.5, color="#4A5568", linestyle=":"); a.axvline(0.5, color="#4A5568", linestyle=":")
    a.set_xlabel("correlação com as vendas")
    for lado in ["top", "right"]:
        a.spines[lado].set_visible(False)
ax[0].set_ylabel("indicadores")
fig.suptitle(titulo, x=0.02, ha="left", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()
```

# Parte 6 · Vida+ Seguros: quem fez o treinamento vende mais?
A seguradora ofereceu um **treinamento opcional** de vendas aos 400 corretores. Quem fez **vendeu bem mais** no semestre seguinte, e o RH quer tornar o treinamento **obrigatório** (cerca de **R\$ 1,2 milhão por ano**, custo hipotético). A pergunta: **o treinamento aumenta as vendas?**

| Coluna | O que é |
|---|---|
| `corretor` | código do corretor |
| `vendas_antes` | índice de vendas no semestre **antes** do treinamento |
| `fez_treinamento` | `True` se o corretor escolheu fazer |
| `vendas_depois` | índice de vendas no semestre **depois** do treinamento |

**O que acontece na célula abaixo**

- Lê a base e mostra as 5 primeiras linhas.

```python
corretores = pd.read_csv(RAIZ + "aula05_c6_corretores.csv")
corretores.head()
```

**O que a diretoria viu: quem fez × quem não fez, depois**

**O que acontece na célula abaixo**

- A tabela mostra, para quem fez e quem não fez, a média das vendas **depois** e o número de corretores.
- `diferenca` compara as duas médias, em %.

```python
depois = corretores.groupby("fez_treinamento")["vendas_depois"].mean()
print(corretores.groupby("fez_treinamento")["vendas_depois"].agg(["mean", "count"]))
print(f"Quem fez vendeu {depois[True] / depois[False] - 1:.1%} a mais")
```

**O que olhar na saída**

A diferença parece grande? Agora pense em **quem escolhe** fazer um treinamento opcional. Esses corretores já eram diferentes **antes**?

**O teste: olhar também o semestre anterior**

**O que acontece na célula abaixo**

- A primeira tabela compara as vendas **antes** e **depois**, para quem fez e quem não fez.
- `crescimento` é quanto cada corretor cresceu, em %; a última linha compara o crescimento médio dos dois grupos.

```python
print(corretores.groupby("fez_treinamento")[["vendas_antes", "vendas_depois"]].mean())
print()
corretores["crescimento"] = (corretores["vendas_depois"] / corretores["vendas_antes"] - 1) * 100
corretores.groupby("fez_treinamento")["crescimento"].mean()
```

**O que olhar na saída**

Quem fez o treinamento já vendia mais **antes**? E o **crescimento** de quem fez foi maior que o de quem não fez? O que sobra da diferença de antes?

**O que acontece na célula abaixo**

- Este é o gráfico para o slide, já no formato do guia: **um gráfico, uma mensagem**; cinza para o contexto e **uma cor** para o que importa; o número escrito no lugar onde é lido; sem grade e sem bordas.
- **Troque o texto de `titulo`** pela sua Big Idea e execute de novo. Para levar ao slide: clique com o botão direito na imagem e copie (ou salve).

```python
titulo = "Escreva aqui a sua Big Idea"

tabela = corretores.groupby("fez_treinamento")[["vendas_antes", "vendas_depois"]].mean()

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(["antes", "depois"], tabela.loc[True], marker="o", color="#0E6FA6", linewidth=3)
ax.plot(["antes", "depois"], tabela.loc[False], marker="o", color="#A9B5C7", linewidth=3)
ax.text(1.05, tabela.loc[True, "vendas_depois"], "fizeram o treinamento", color="#0E6FA6", va="center", fontsize=12, fontweight="bold")
ax.text(1.05, tabela.loc[False, "vendas_depois"], "não fizeram", color="#4A5568", va="center", fontsize=12)
ax.set_title(titulo, loc="left", fontsize=14, fontweight="bold")
ax.set_ylabel("índice de vendas"); ax.set_xlim(-0.2, 1.8)
for lado in ["top", "right"]:
    ax.spines[lado].set_visible(False)
plt.show()
```
