## 8. Verificador
Execute antes de **cada** entrega, com a variável `ETAPA` correta. O corretor lê a saída desta célula.

```python
import json

def verificar(etapa="final"):
    ok = True
    g = globals()
    if not MATRICULA:
        print("⚠️ MATRICULA vazia"); ok = False
    m1 = g.get("MARCO1", {})
    for k in ("pergunta", "tecnica", "dados_necessarios", "medida_de_sucesso"):
        if not str(m1.get(k, "")).strip():
            print(f"⚠️ MARCO1['{k}'] vazio"); ok = False
    if etapa in ("marco2", "marco3", "final"):
        d = g.get("DICIONARIO", {})
        if len(d) < 10:
            print(f"⚠️ DICIONARIO: descreva pelo menos 10 colunas (há {len(d)})"); ok = False
    if etapa in ("marco3", "final"):
        r = g.get("RESPOSTAS_FINAL", {})
        faltam = [k for k, v in r.items() if v is None or (isinstance(v, str) and not v.strip())]
        if etapa == "final" and faltam:
            print("⚠️ Respostas vazias:", ", ".join(faltam)); ok = False
        for k in ("F7_pct_atraso", "F11_pct_sem_nota_atrasados", "F11_pct_sem_nota_no_prazo"):
            try:
                if 0 < float(r.get(k)) < 1:
                    print(f"⚠️ {k}: use porcentagem de 0 a 100"); ok = False
            except (TypeError, ValueError):
                pass
    if etapa == "final":
        n = len(str(g.get("MEMORANDO", "")).split())
        if n < 120:
            print(f"⚠️ MEMORANDO muito curto ({n} palavras)"); ok = False
        if n > 400:
            print(f"⚠️ MEMORANDO com {n} palavras (máximo 400)"); ok = False
        if "Descreva aqui" in str(g.get("DECLARACAO_IA", "")):
            print("⚠️ Preencha a DECLARACAO_IA"); ok = False
    print(("✅ Formato OK para a etapa " + etapa) if ok else "Corrija os itens acima antes de entregar.")
    limpar = lambda x: {k: (float(v) if hasattr(v, "item") else v) for k, v in x.items()} if isinstance(x, dict) else x
    pacote = {"MATRICULA": MATRICULA, "ETAPA": etapa, "MARCO1": m1, "DICIONARIO": g.get("DICIONARIO", {}),
              "RESPOSTAS_FINAL": limpar(g.get("RESPOSTAS_FINAL", {})), "MEMORANDO": g.get("MEMORANDO", ""),
              "DECLARACAO_IA": g.get("DECLARACAO_IA", "")}
    print("TF_JSON=" + json.dumps(pacote, ensure_ascii=False, default=str))

verificar(ETAPA)
```
