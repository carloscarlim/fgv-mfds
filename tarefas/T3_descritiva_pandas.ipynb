## Verificador de formato
Execute antes de baixar o arquivo. O corretor lê a saída desta célula.

```python
import json

def verificar(r):
    ok = True
    for k in ["Q1_n_pedidos", "Q2_pct_entregues", "Q3_prazo_medio", "Q4_prazo_mediana", "Q5_pct_atraso", "Q7_nota_atrasados", "Q8_nota_no_prazo"]:
        v = r.get(k)
        try:
            float(v)
        except (TypeError, ValueError):
            print(f"⚠️ {k}: precisa ser um número"); ok = False
    for k in ["Q6_uf_maior_atraso", "Q9_categoria_maior_faturamento", "Q10_mensagem_diretor"]:
        if not str(r.get(k, "")).strip():
            print(f"⚠️ {k}: vazio"); ok = False
    for k in ["Q2_pct_entregues", "Q5_pct_atraso"]:
        try:
            if 0 < float(r[k]) <= 1:
                print(f"⚠️ {k}: parece estar entre 0 e 1. Use porcentagem de 0 a 100."); ok = False
        except (TypeError, ValueError, KeyError):
            pass
    n = len(str(r.get("Q10_mensagem_diretor", "")).split())
    if n > 80:
        print(f"⚠️ Q10: {n} palavras (máximo 80)"); ok = False
    if not MATRICULA:
        print("⚠️ MATRICULA vazia"); ok = False
    print("✅ Formato OK. Pode entregar." if ok else "Corrija os itens acima antes de entregar.")
    limpo = {k: (float(v) if hasattr(v, "item") else v) for k, v in r.items()}
    print("RESPOSTAS_JSON=" + json.dumps({"MATRICULA": MATRICULA, **limpo}, ensure_ascii=False, default=str))

verificar(RESPOSTAS)
```
