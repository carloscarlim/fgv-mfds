## Seu sorteio
Não altere esta célula. Ela define o seu estado (Parte A) e os seus números do caso (Parte B).

```python
import random
import re

UFS = ["AC", "AL", "AM", "AP", "BA", "CE", "DF", "ES", "GO", "MA", "MG", "MS", "MT", "PA", "PB",
       "PE", "PI", "PR", "RJ", "RN", "RO", "RR", "RS", "SC", "SE", "SP", "TO"]


def sorteio(matricula):
    digitos = re.sub(r"\D", "", str(matricula))
    if not digitos:
        raise ValueError("Matrícula inválida: use o seu nome de usuário do ECLASS, por exemplo A12345678.")
    semente = int(digitos)
    rng = random.Random(semente)
    return {
        "uf": UFS[semente % 27],
        "volume": rng.randrange(15000, 30001, 1000),
        "acerto": rng.randrange(55, 86),
        "ganho": rng.randrange(400, 901, 50),
        "perda": rng.randrange(250, 601, 25),
        "custo_decisao": rng.randrange(5, 21) / 10,
        "custo_fixo": rng.randrange(20, 51) * 100000,
        "risco": rng.randrange(2, 11) * 50000,
    }

MEU = sorteio(MATRICULA)
UF = MEU['uf']
print(f'Seu estado na Parte A: {UF}')
print('Seus números na Parte B:')
print(f"  Volume de inspeções apoiadas por ano: {MEU['volume']:,}".replace(',', '.'))
print(f"  Taxa de acerto do modelo: {MEU['acerto']}%")
print(f"  Ganho por inspeção certa (energia recuperada): R$ {MEU['ganho']:,}".replace(',', '.'))
print(f"  Perda por inspeção errada (viagem perdida): R$ {MEU['perda']:,}".replace(',', '.'))
print(f"  Custo de rodar o modelo, por decisão: R$ {MEU['custo_decisao']:.2f}".replace('.', ','))
print(f"  Custo fixo anual: R$ {MEU['custo_fixo']:,}".replace(',', '.'))
print(f"  Risco esperado anual: R$ {MEU['risco']:,}".replace(',', '.'))
```
